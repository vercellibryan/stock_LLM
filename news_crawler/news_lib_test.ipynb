## Crawl functions

```python
# company_name: str, ticker: str, start_date: date, end_date: date, sites: list[str]
FINANCE_PUBLISHERS = ["finance.yahoo.com", "benzinga.com"]
query_url = nl.query_builder("NVIDIA", "NVDA", date(2026, 6, 1), date(2026, 6, 3), FINANCE_PUBLISHERS)
```

```python
# link: str, start_date: date, end_date: date, sites: list[str], limit:int = 0
crawl_news = nl.news_filter(query_url, date(2026, 6, 1), date(2026, 6, 3), FINANCE_PUBLISHERS)
crawl_news[0]
```

```python
crawled_news = nl.crawl_full_text(crawl_news[:3])
crawled_news
```

## Upsert Articles

```python
nl.upsert_articles("NVDA", crawl_news)
```
